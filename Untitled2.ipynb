2. Функція, яка приймає словник, старий ключ та новий ключ, і повертає словник зі зміненим ключем.

```python
def format_name(first_name, patronymic, last_name):
    return f"{first_name[0]}. {patronymic[0]}. {last_name}"

# Тестування за допомогою assert (як у прикладі з методички)
assert format_name("Володимир", "Олександрович", "Зеленський") == "В. О. Зеленський"
assert format_name("Тарас", "Григорович", "Шевченко") == "Т. Г. Шевченко"
assert format_name("Тест", "2", "Іванів") == "Т. 2. Іванів"
assert format_name("3", "2", "Бондар") == "3. 2. Бондар"
assert format_name("Максим", "Кост", "Тестовий") == "М. К. Тестовий"

print("Завдання 1.1: Усі тести пройдено успішно!")
```

```python
def change_key(dictionary, old_key, new_key):
    # Якщо старий ключ є у словнику, замінюємо його
    if old_key in dictionary:
        dictionary[new_key] = dictionary.pop(old_key)
    return dictionary

# Тестування за допомогою assert
dict1 = {'1': 1, 'total': 55}
assert change_key(dict1.copy(), 'total', 'sum') == {'1': 1, 'sum': 55}

dict2 = {'a': 100, 'b': 200}
assert change_key(dict2.copy(), 'a', 'alpha') == {'alpha': 100, 'b': 200}

dict3 = {'x': 10, 'y': 20}
assert change_key(dict3.copy(), 'z', 'gamma') == {'x': 10, 'y': 20} # ключа немає, словник не змінюється

dict4 = {'hello': 'world'}
assert change_key(dict4.copy(), 'hello', 'hi') == {'hi': 'world'}

print("Завдання 1.2: Усі тести пройдено успішно!")
```
